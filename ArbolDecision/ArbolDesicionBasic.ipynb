```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier, plot_tree
import matplotlib.pyplot as plt

# 1. Load Fruit Data
dataset = pd.read_csv("../Data/fruits.csv")

# 2. Train a Decision Tree Classifier
X = dataset[["Size", "Weight"]]
y = dataset["Label"]
clf = DecisionTreeClassifier(max_depth=3)
clf.fit(X, y)

# 3. Plot the Decision Tree
plt.figure(figsize=(12, 8))
plot_tree(clf, feature_names=["Size", "Weight"], class_names=clf.classes_, filled=True)
plt.show()
```

```python

test_data = pd.DataFrame({"Size": [160], "Weight": [8]})
clf.predict(test_data[["Size", "Weight"]])[0] # -> lemon
```
